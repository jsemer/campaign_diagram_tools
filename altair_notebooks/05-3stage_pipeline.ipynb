## 3-Stage Pipeline

```python
kernel3a = Kernel(name='EinsumA',
                  start=0,
                  duration=5,
                  compute_util=0.3,
                  bw_util=0.15)


kernel3b = Kernel(name='EinsumB',
                  start=kernel3a.end,
                  duration=10,
                  compute_util=0.2,
                  bw_util=0.6)


kernel3c = Kernel(name='EinsumC',
                  start=kernel3b.end,
                  duration=5,
                  compute_util=0.5,   # was 0.2
                  bw_util=0.25)

original_cascade = Cascade(name="Three-step Cascade",
                           kernels=[kernel3a, kernel3b, kernel3c])
```

#### Untiled cascade 

```python
campaign_diagram = CampaignDiagram(original_cascade)
CampaignDiagram(original_cascade).draw().interactive().show()
CampaignDiagram(original_cascade).draw(dual_mode=True).interactive()
```

#### Tiled Cascade
Tile the cascade, but do not pipeline yet.

```python
num_of_tiles = 4
```

```python
tiled_cascade = original_cascade.tile(num_of_tiles)
CampaignDiagram(tiled_cascade).draw().interactive().show()
CampaignDiagram(tiled_cascade).draw(dual_mode=True).interactive()
```

#### Pipeline!

Apply a 3-stage pipeline (since we want all three Einsums pipelined). 



```python
pipelined_cascade = tiled_cascade.pipeline(stages=3) # the default is 2, so pass "3" to indicate the number of stages.

CampaignDiagram(pipelined_cascade).draw().interactive().show()
CampaignDiagram(pipelined_cascade).draw(dual_mode=True).interactive()
```

#### What happens if we throttle as well?
It should look the same since no group overflowed its resources.

```python
pipelined_throttle_cascade = pipelined_cascade.throttle() 

CampaignDiagram(pipelined_throttle_cascade).draw().interactive().show()
CampaignDiagram(pipelined_throttle_cascade).draw(dual_mode=True).interactive()
```

#### Dilate in Time 

We can dilate in time to reduce burstiness (<mark>TODO: is this the reason </mark>)

```python
# Pipelined diagram (dilated)

pipelined_cascade2 = tiled_cascade.pipeline(stages=3, spread=True)
    
CampaignDiagram(pipelined_cascade2).draw().interactive().show()
CampaignDiagram(pipelined_cascade2).draw(dual_mode=True).interactive()
```
