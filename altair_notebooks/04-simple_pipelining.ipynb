#### Create a 2-step Cascade:

```python
num_of_tiles = 4

kernel2a = Kernel(name='EinsumA',
                  start=0,
                  duration=5,
                  compute_util=0.8,
                  bw_util=0.2)


kernel2b = Kernel(name='EinsumB',
                  start=kernel2a.end,
                  duration=10,
                  compute_util=0.2,
                  bw_util=0.6)

original_cascade = Cascade(name="Two-step Cascade",
                           kernels=[kernel2a, kernel2b])
```

```python
campaign_diagram = CampaignDiagram(original_cascade)
CampaignDiagram(original_cascade).draw().interactive().show()
CampaignDiagram(original_cascade).draw(dual_mode=True).interactive()
```

Suppose the cascade runs such that a bit of Einsum A runs, then a bit of Einsum B, repeating 4 times.
Additionally, we assume a tile of B is dependent on the previous tile of A.

```python
tiled_cascade = original_cascade.tile(num_of_tiles)
CampaignDiagram(tiled_cascade).draw().interactive().show()
CampaignDiagram(tiled_cascade).draw(dual_mode=True).interactive()
```

Since a previous tile of orange is independent of the next tile of blue, we can overap their execution (pipeline parallelism).  
We do this by calling `cascade.pipeline()`
This function takes a *tiled* cascade and creates (by default) a 2-deep pipeline. 



```python
pipelined_cascade = tiled_cascade.pipeline()

CampaignDiagram(pipelined_cascade).draw().interactive().show()
CampaignDiagram(pipelined_cascade).draw(dual_mode=True).interactive()
```

Note that the above cascade is bursty (Einsum A runs with Einsum B at peak utilization for a short amount of time).

To reduce this burstiness, we can apply *spread*, which, within a pipelined group, stretches or *dilates* the shorter Einsum in time to end at the same time as the bottleneck Einsum. Call `cascade.pipeline(spread=True)`

Note that this will now reduce the compute resource utilization and memory bandwidth utilization of the stretched Einsum:

```python
pipelined_and_spead_cascade = tiled_cascade.pipeline(spread=True)

CampaignDiagram(pipelined_and_spead_cascade).draw().interactive().show()
CampaignDiagram(pipelined_and_spead_cascade).draw(dual_mode=True).interactive()
```
