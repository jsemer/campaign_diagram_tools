## Tiling - Option 1

Suppose we have two kernels and we want to show the campaign diagram for a partitioned mapping, where some of Einsum A executes, then some of Einsum B executes.

First, define the duration and utilizations for a tile.

```python
# Show replicated sequence

kernel2a = Kernel(name='EinsumA',
                  duration=3,
                  compute_util=0.7,
                  bw_util=0.25)

kernel2b = Kernel(name='EinsumB',
                  duration=10,
                  compute_util=0.2,
                  bw_util=0.9)


```

Then duplicate the kernels and create a cascade!

```python
kernels2 = [kernel.clone() for kernel in 4*[kernel2a, kernel2b]]

cascade2 = Cascade(name="Sequential Cascade",
                   kernels=kernels2,
                   sequential=True)

cascade2.pretty_print()
```

```python
campaign_diagram = CampaignDiagram(cascade2)
campaign_diagram.draw().interactive()
```

In dual mode:

```python
campaign_diagram = CampaignDiagram(cascade2)
campaign_diagram.draw(dual_mode=False).interactive().show()
campaign_diagram.draw(dual_mode=True).interactive()
```

## Tiling - Option 2

We can also tile by calling `cascade.tile(num_of_tiles)`.
This splits a given cascade into chunks such that the overall runtime is the same as the unsplit cascade.


```python
# create the cascade
cascade2 = Cascade(name="Sequential Cascade",
                   kernels=[kernel2a, kernel2b],
                   sequential=True)
num_of_tiles = 4
# tile by calling `tile`
tiled_cascade = cascade2.tile(num_of_tiles)
CampaignDiagram(tiled_cascade).draw().interactive().show()

campaign_diagram = CampaignDiagram(tiled_cascade)
campaign_diagram.draw(dual_mode=True).interactive().show()
```
